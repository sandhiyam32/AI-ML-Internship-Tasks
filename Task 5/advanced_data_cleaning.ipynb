```python


import pandas as pd
import numpy as np

# ---------------------------------------------------------
# 1. Load the dataset
# ---------------------------------------------------------

file_name = "customer_data.csv"

df = pd.read_csv(file_name)

print("ORIGINAL DATASET")
print("-" * 50)

print("Dataset Shape:", df.shape)
print("\nFirst 5 Records:")
print(df.head())

# ---------------------------------------------------------
# 2. Check data types
# ---------------------------------------------------------

print("\nDATA TYPES")
print("-" * 50)
print(df.dtypes)

# ---------------------------------------------------------
# 3. Check missing values
# ---------------------------------------------------------

print("\nMISSING VALUES BEFORE CLEANING")
print("-" * 50)
print(df.isnull().sum())

# ---------------------------------------------------------
# 4. Remove duplicate records
# ---------------------------------------------------------

duplicate_count = df.duplicated().sum()

print("\nDUPLICATE RECORDS:", duplicate_count)

df = df.drop_duplicates()

# ---------------------------------------------------------
# 5. Clean column names
# ---------------------------------------------------------

df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# ---------------------------------------------------------
# 6. Clean text columns
# ---------------------------------------------------------

text_columns = ["name", "gender", "city"]

for column in text_columns:
    df[column] = df[column].astype("string").str.strip()

# ---------------------------------------------------------
# 7. Standardize Gender values
# ---------------------------------------------------------

gender_mapping = {
    "m": "Male",
    "male": "Male",
    "MALE": "Male",
    "f": "Female",
    "female": "Female",
    "FEMALE": "Female"
}

df["gender"] = df["gender"].replace(gender_mapping)

# ---------------------------------------------------------
# 8. Standardize City values
# ---------------------------------------------------------

df["city"] = df["city"].str.title()

# ---------------------------------------------------------
# 9. Convert Age into numeric format
# ---------------------------------------------------------

df["age"] = pd.to_numeric(df["age"], errors="coerce")

# Replace missing age with median
df["age"] = df["age"].fillna(df["age"].median())

# ---------------------------------------------------------
# 10. Clean Salary values
# ---------------------------------------------------------

df["salary"] = (
    df["salary"]
    .astype("string")
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["salary"] = pd.to_numeric(df["salary"], errors="coerce")

# Replace missing salary with median
df["salary"] = df["salary"].fillna(df["salary"].median())

# ---------------------------------------------------------
# 11. Clean Email addresses
# ---------------------------------------------------------

df["email"] = (
    df["email"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# ---------------------------------------------------------
# 12. Clean Phone numbers
# ---------------------------------------------------------

df["phone"] = (
    df["phone"]
    .astype("string")
    .str.replace(r"\D", "", regex=True)
)

# ---------------------------------------------------------
# 13. Handle missing categorical values
# ---------------------------------------------------------

df["gender"] = df["gender"].fillna(df["gender"].mode()[0])
df["city"] = df["city"].fillna(df["city"].mode()[0])

# ---------------------------------------------------------
# 14. Check missing values after cleaning
# ---------------------------------------------------------

print("\nMISSING VALUES AFTER CLEANING")
print("-" * 50)
print(df.isnull().sum())

# ---------------------------------------------------------
# 15. Check duplicate records after cleaning
# ---------------------------------------------------------

print("\nDUPLICATES AFTER CLEANING:", df.duplicated().sum())

# ---------------------------------------------------------
# 16. Save cleaned dataset
# ---------------------------------------------------------

df.to_csv("cleaned_customer_data.csv", index=False)

# ---------------------------------------------------------
# 17. Display cleaned dataset
# ---------------------------------------------------------

print("\nCLEANED DATASET")
print("-" * 50)
print(df.head())

print("\nFinal Dataset Shape:", df.shape)

print("\nData cleaning completed successfully!")
print("Cleaned file saved as: cleaned_customer_data.csv")
```
