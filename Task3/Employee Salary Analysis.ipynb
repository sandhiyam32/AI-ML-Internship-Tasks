```python
import pandas as pd

data = {
    'Employee_ID': ['E001', 'E002', 'E003', 'E004', 'E005', 'E006', 'E007'],
    'Salary': [45000, 52000, 68000, 55000, 74000, 48000, 61000]
}
df = pd.DataFrame(data)

total_salary = df['Salary'].sum()
average_salary = df['Salary'].mean()
highest_salary = df['Salary'].max()
lowest_salary = df['Salary'].min()


print("Salary Analysis Result")
print(f"Total Salary Payout: ${total_salary:,.2f}")
print(f"Average Salary:      ${average_salary:,.2f}")
print(f"Highest Salary:      ${highest_salary:,.2f}")
print(f"Lowest Salary:       ${lowest_salary:,.2f}")


print("\n--- Quick Statistical Summary ---")
print(df['Salary'].describe())
```
