```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# ---------------------------------------------------------
# 1. Load the Dataset
# ---------------------------------------------------------

df = pd.read_csv("Titanic.csv")

print("ORIGINAL DATASET")
print("-" * 50)
print(df.head())

# ---------------------------------------------------------
# 2. Display Original Categorical Columns
# ---------------------------------------------------------

print("\nORIGINAL CATEGORICAL DATA")
print("-" * 50)
print(df[["Sex", "Embarked"]])

# ---------------------------------------------------------
# 3. Create a Copy of the Dataset
# ---------------------------------------------------------

encoded_df = df.copy()

# ---------------------------------------------------------
# 4. Apply Label Encoding to Sex
# ---------------------------------------------------------

sex_encoder = LabelEncoder()

encoded_df["Sex_Encoded"] = sex_encoder.fit_transform(
    encoded_df["Sex"]
)

# ---------------------------------------------------------
# 5. Apply Label Encoding to Embarked
# ---------------------------------------------------------

embarked_encoder = LabelEncoder()

encoded_df["Embarked_Encoded"] = embarked_encoder.fit_transform(
    encoded_df["Embarked"]
)

# ---------------------------------------------------------
# 6. Display Encoded Data
# ---------------------------------------------------------

print("\nDATA AFTER LABEL ENCODING")
print("-" * 50)

print(
    encoded_df[
        ["Sex", "Sex_Encoded",
         "Embarked", "Embarked_Encoded"]
    ]
)

# ---------------------------------------------------------
# 7. Display Encoding Mapping for Sex
# ---------------------------------------------------------

print("\nSEX ENCODING MAPPING")
print("-" * 50)

for label, value in zip(
    sex_encoder.classes_,
    sex_encoder.transform(sex_encoder.classes_)
):
    print(label, "->", value)

# ---------------------------------------------------------
# 8. Display Encoding Mapping for Embarked
# ---------------------------------------------------------

print("\nEMBARKED ENCODING MAPPING")
print("-" * 50)

for label, value in zip(
    embarked_encoder.classes_,
    embarked_encoder.transform(embarked_encoder.classes_)
):
    print(label, "->", value)

# ---------------------------------------------------------
# 9. Save Encoded Dataset
# ---------------------------------------------------------

encoded_df.to_csv(
    "Task_7_Label_Encoded_Output.csv",
    index=False
)

print("\nEncoded dataset saved successfully!")
```
