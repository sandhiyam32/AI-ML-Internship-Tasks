```python

marks = [78, 85, 92, 67, 88, 74, 95, 81, 69, 90]

total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)
print("Student Marks Analysis")
print("Marks:", marks)
print("Total Marks:", total)
print("Average Marks:", round(average, 2))
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
```
