```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing

data = fetch_california_housing(as_frame=True)
df = data.frame
print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
print(df.head())
features = ["MedInc", "HouseAge", "AveRooms", "Population", "AveOccup"]
print("\n--- Feature Distribution Analysis ---")

for feature in features:

    data = df[feature].dropna()

    mean = data.mean()
    median = data.median()
    skewness = data.skew()

    print(f"\nFeature: {feature}")
    print(f"Mean     : {mean:.2f}")
    print(f"Median   : {median:.2f}")
    print(f"Skewness : {skewness:.2f}")

    # Identify skewness
    if skewness > 1:
        print("Observation: Strongly right-skewed")
    elif skewness > 0.5:
        print("Observation: Moderately right-skewed")
    elif skewness < -1:
        print("Observation: Strongly left-skewed")
    elif skewness < -0.5:
        print("Observation: Moderately left-skewed")
    else:
        print("Observation: Approximately symmetric")

    
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = data[(data < lower) | (data > upper)]

    print(f"Potential Outliers: {len(outliers)}")

 
    plt.figure(figsize=(8, 5))
    sns.histplot(data, kde=True, bins=30)

    plt.axvline(mean, linestyle="--", label="Mean")
    plt.axvline(median, linestyle="-.", label="Median")

    plt.title(f"Distribution of {feature}")
    plt.xlabel(feature)
    plt.ylabel("Frequency")
    plt.legend()
    plt.show()

   
    plt.figure(figsize=(8, 3))
    sns.boxplot(x=data)

    plt.title(f"Box Plot of {feature}")
    plt.xlabel(feature)
    plt.show()
```
