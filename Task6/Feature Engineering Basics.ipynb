Feature Engineering Basics

```python
import pandas as pd
import numpy as np

# ---------------------------------------------------------
# 1. Import Titanic Dataset
# ---------------------------------------------------------

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("ORIGINAL TITANIC DATASET")
print("-" * 50)
print(df.head())

print("\nDataset Shape:", df.shape)

# ---------------------------------------------------------
# 2. Feature 1 - Family Size
# ---------------------------------------------------------

df["Family_Size"] = df["SibSp"] + df["Parch"] + 1

# ---------------------------------------------------------
# 3. Feature 2 - Is Alone
# ---------------------------------------------------------

df["Is_Alone"] = np.where(df["Family_Size"] == 1, 1, 0)

# ---------------------------------------------------------
# 4. Feature 3 - Fare Per Person
# ---------------------------------------------------------

df["Fare_Per_Person"] = (
    df["Fare"] / df["Family_Size"]
)

# ---------------------------------------------------------
# 5. Feature 4 - Age Group
# ---------------------------------------------------------

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 12, 18, 60, 100],
    labels=["Child", "Teenager", "Adult", "Senior"]
)

# ---------------------------------------------------------
# 6. Display Engineered Features
# ---------------------------------------------------------

print("\nENGINEERED FEATURES")
print("-" * 50)

print(
    df[
        [
            "Name",
            "Age",
            "SibSp",
            "Parch",
            "Fare",
            "Family_Size",
            "Is_Alone",
            "Fare_Per_Person",
            "Age_Group"
        ]
    ].head(10)
)

# ---------------------------------------------------------
# 7. Explanation
# ---------------------------------------------------------

print("\nFEATURE EXPLANATION")
print("-" * 50)

print("1. Family_Size = SibSp + Parch + 1")
print("   Shows the total number of family members travelling.")

print("\n2. Is_Alone")
print("   1 means the passenger travelled alone.")
print("   0 means the passenger travelled with family.")

print("\n3. Fare_Per_Person = Fare / Family_Size")
print("   Shows the approximate fare paid per family member.")

print("\n4. Age_Group")
print("   Converts numerical age into meaningful categories.")

# ---------------------------------------------------------
# 8. Compare Original and Engineered Features
# ---------------------------------------------------------

print("\nORIGINAL VS ENGINEERED FEATURES")
print("-" * 50)

comparison = df[
    [
        "Age",
        "SibSp",
        "Parch",
        "Fare",
        "Family_Size",
        "Is_Alone",
        "Fare_Per_Person",
        "Age_Group"
    ]
]

print(comparison.head(10))

# ---------------------------------------------------------
# 9. Save the Engineered Dataset
# ---------------------------------------------------------

df.to_csv("titanic_engineered.csv", index=False)

print("\nFeature engineering completed successfully!")
print("Saved file: titanic_engineered.csv")
```
