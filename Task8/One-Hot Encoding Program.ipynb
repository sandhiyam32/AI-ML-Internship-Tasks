```python
import pandas as pd

# 1. Load the dataset
df = pd.read_csv("One_Hot_Encoding_Dataset.csv")

print("ORIGINAL DATA")
print("-" * 50)
print(df.head())

# 2. Select categorical columns
categorical_columns = ["Gender", "Department", "City"]

print("\nCATEGORICAL COLUMNS")
print("-" * 50)
print(df[categorical_columns].head())

# 3. Apply One-Hot Encoding
encoded_df = pd.get_dummies(
    df,
    columns=categorical_columns,
    dtype=int
)

# 4. Display encoded data
print("\nDATA AFTER ONE-HOT ENCODING")
print("-" * 50)
print(encoded_df.head())

# 5. Display number of columns
print("\nNumber of columns before encoding:", len(df.columns))
print("Number of columns after encoding:", len(encoded_df.columns))

# 6. Save the encoded dataset
encoded_df.to_csv(
    "Task_8_One_Hot_Encoded_Output.csv",
    index=False
)
```
