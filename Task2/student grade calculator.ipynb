```python
students = [
    ("Arun", 92),
    ("Priya", 85),
    ("Rahul", 76),
    ("Meena", 64),
    ("Kavin", 48),
    ("Divya", 91)
]

for name, marks in students:

    if marks >= 90:
        grade = "A"
    elif marks >= 80:
        grade = "B"
    elif marks >= 70:
        grade = "C"
    elif marks >= 60:
        grade = "D"
    else:
        grade = "F"

    print(f"{name}: {marks} → Grade {grade}")
```
