# Task 23 – K-Means Clustering

This notebook uses a small Mall Customer-style dataset to group customers based on annual income and spending score.

## 1. Import libraries and load the dataset
Keep the CSV file in the same folder as this notebook.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from tkinter import Tk, filedialog

# 1. Select the customer dataset
root = Tk()
root.withdraw()

file_path = filedialog.askopenfilename(
    title="Select Customer Dataset",
    filetypes=[("CSV Files", "*.csv")]
)

root.destroy()

# 2. Load the dataset
df = pd.read_csv(file_path)

print("Original Customer Dataset")
print("-" * 50)
display(df.head())

print("Dataset shape:", df.shape)
```

## 2. Select and scale the features

```python
features = df[['AnnualIncome_kUSD', 'SpendingScore']]

scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
```

## 3. Train the K-Means model

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(scaled_features)

print('Cluster assignments')
display(df[['CustomerID', 'AnnualIncome_kUSD', 'SpendingScore', 'Cluster']].head(10))
```

## 4. Visualize the clusters

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['AnnualIncome_kUSD'], df['SpendingScore'], c=df['Cluster'], cmap='viridis', s=70)
plt.xlabel('Annual Income (thousands of dollars)')
plt.ylabel('Spending Score')
plt.title('Customer Groups using K-Means')
plt.grid(True, alpha=0.3)
plt.show()
```

## 5. Interpret the clusters

```python
print('Average values for each cluster')
display(df.groupby('Cluster')[['AnnualIncome_kUSD', 'SpendingScore']].mean().round(2))

print('Number of customers in each cluster')
display(df['Cluster'].value_counts().sort_index())
```

## Conclusion

K-Means grouped customers into three clusters based on annual income and spending score. The summary table helps compare the average values in each group. Cluster numbers are labels only; they do not represent a ranking.
