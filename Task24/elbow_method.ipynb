# Task 24 – Elbow Method for K-Means

Compare K-Means inertia across several cluster counts, plot the elbow curve, and select a suitable K for the customer dataset.

## 1. Import libraries and load the dataset

Run Jupyter with the Task24 folder as its working folder, and keep `Mall_Customers.csv` beside this notebook.


```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

customers = pd.read_csv("Mall_Customers.csv")
display(customers.head())
print(f"Dataset shape: {customers.shape}")
```

## 2. Select and scale the features

`CustomerID` is only an identifier, so clustering uses annual income and spending score. Standardizing prevents the different feature scales from dominating the distance calculations.

```python
features = customers[['AnnualIncome_kUSD', 'SpendingScore']]
scaled_features = StandardScaler().fit_transform(features)
print(f'Features used: {list(features.columns)}')
```

## 3. Fit K-Means for multiple K values

Inertia is the sum of squared distances from samples to their assigned cluster centers. Lower inertia indicates tighter clusters, and it typically decreases as K increases.

```python
k_values = range(1, 11)
inertias = []

for k in k_values:
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(scaled_features)
    inertias.append(model.inertia_)

inertia_results = pd.DataFrame({'K': list(k_values), 'Inertia': inertias})
display(inertia_results.round(3))
```

## 4. Plot the elbow curve

Look for the bend where adding another cluster starts to provide less improvement.

```python
selected_k = 3

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(list(k_values), inertias, marker='o')
ax.axvline(selected_k, color='tab:red', linestyle='--', label=f'Selected K = {selected_k}')
ax.set_title('Elbow Method for Mall Customer Data')
ax.set_xlabel('Number of clusters (K)')
ax.set_ylabel('Inertia')
ax.set_xticks(list(k_values))
ax.grid(True, alpha=0.3)
ax.legend()
fig.tight_layout()

plot_path = Path('elbow_curve.png')
fig.savefig(plot_path, dpi=150)
print(f'Elbow curve saved to: {plot_path}')
plt.show()
plt.close(fig)
```

## 5. Selected K and explanation

The selected value is **K=3**. On the standardized features, inertia falls sharply from about **32.554** at K=2 to **3.364** at K=3, then decreases more gradually to about **2.680** at K=4. This bend suggests that additional clusters after K=3 provide diminishing improvement for this sample. The Elbow Method is a practical guide rather than proof that one K is universally best.
